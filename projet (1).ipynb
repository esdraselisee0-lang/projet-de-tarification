```r
freq_par <- function(data, var) {
  data %>%
    group_by({{ var }}) %>%
    summarise(
      exposition = sum(Exposure),
      sinistres  = sum(ClaimNb),
      frequence  = sinistres / exposition,
      .groups = "drop"
    )
}
```

```r
glm_freq <- glm(
  ClaimNb ~ DrivAgeClass + VehAgeClass + VehPower + VehGas +
    Area + BonusMalus + logDensity + VehBrand +
    offset(log(Exposure)),
  family = poisson(link = "log"),
  data = train
)
```

```r
glm_sev <- glm(
  CoutMoyen ~ DrivAgeClass + VehAgeClass + VehPower + VehGas +
    Area + logDensity,
  family = Gamma(link = "log"),
  weights = ClaimNb,
  data = train_sev
)
summary(glm_sev)
```

```r
calc_prime_pure <- function(data) {
  d1 <- data %>% mutate(Exposure = 1)  # on calcule pour 1 an
  f  <- predict(glm_freq, newdata = d1, type = "response")
  s  <- predict(glm_sev,  newdata = d1, type = "response")
  tibble(freq_annuelle = f, cout_moyen = s, prime_pure = f * s)
}
```

```r
gini <- function(pred, obs, expo) {
  o <- order(pred)
  cum_expo <- cumsum(expo[o]) / sum(expo)
  cum_obs  <- cumsum(obs[o])  / sum(obs)
  # aire sous la courbe par trapezes ; Gini = 1 - 2 * aire
  aire <- sum(diff(c(0, cum_expo)) * (cum_obs + c(0, head(cum_obs, -1))) / 2)
  1 - 2 * aire
}
gini(pred_test$prime_pure, pred_test$ClaimAmount, pred_test$Exposure)
```

```r
taux_frais <- 0.20   # hypothese : 20 % de frais
elast_a    <- 1.0    # parametres de la loi logistique (hypotheses)
elast_b    <- 8.0

# Prix de marche fictif : prime pure x 1.25 (a justifier dans le rapport)
pred_test <- pred_test %>% mutate(prix_marche = prime_pure * 1.25)

profit_attendu <- function(marge, data) {
  prix  <- data$prime_pure / (1 - taux_frais - marge)
  ratio <- prix / data$prix_marche
  p_sub <- plogis(elast_a - elast_b * (ratio - 1))   # proba de souscription
  sum(p_sub * (prix - data$prime_pure - taux_frais * prix))
}

marges <- seq(0, 0.40, by = 0.01)
resultats <- tibble(
  marge  = marges,
  profit = sapply(marges, profit_attendu, data = pred_test)
)

marge_opt <- resultats$marge[which.max(resultats$profit)]
marge_opt

ggplot(resultats, aes(marge, profit)) +
  geom_line() +
  geom_vline(xintercept = marge_opt, linetype = "dashed", colour = "red") +
  labs(title = "Profit attendu selon la marge cible",
       x = "Marge cible", y = "Profit attendu")

# Variante avec optimize() de R
optimize(function(m) -profit_attendu(m, pred_test),
         interval = c(0, 0.40))
# Avant de calculer profit_attendu, filtrez les NA
pred_test <- pred_test %>% filter(!is.na(prime_pure))
```

# Présentation du projet

Ce projet construit une prime d'assurance automobile par le modèle **fréquence × coût moyen**, puis cherche la marge qui maximise le profit attendu de l'assureur.

- **Fréquence** : nombre de sinistres par an, modélisé par une régression de Poisson avec lien log et décalage (`offset`) sur l'exposition.
- **Coût moyen** : montant moyen d'un sinistre, modélisé par une régression Gamma avec lien log, pondérée par le nombre de sinistres.
- **Prime pure** : produit des deux, calculée pour un an de garantie :

$$
\text{prime pure}(x) = \hat{f}(x) \times \hat{s}(x)
$$

où $\hat{f}(x)$ est la fréquence annuelle prédite et $\hat{s}(x)$ le coût moyen prédit pour le profil $x$.

# Justification des données

**Jeu de données (freMTPL2freq et freMTPL2sev, package CASdatasets).** Ce sont deux tables réelles d'un portefeuille français d'assurance auto responsabilité civile, très utilisées en actuariat pour illustrer la tarification. La table de fréquence contient 677 991 contrats avec leurs caractéristiques (âge du conducteur, âge et puissance du véhicule, carburant, zone, densité, bonus-malus, marque) et le nombre de sinistres. La table de sévérité contient le montant de chaque sinistre. Le volume est assez grand pour estimer des effets fins sans surajustement.

**Jointure des deux tables.** Les montants de `sev` sont d'abord agrégés par contrat (`IDpol`), puis rattachés à `freq`. Un contrat sans sinistre reçoit un montant de 0 (`replace_na`). On obtient une ligne par contrat avec nombre de sinistres, exposition et charge totale.

**Un seul portefeuille, une seule période.** Les données ne donnent ni inflation, ni évolution du portefeuille dans le temps. Les primes calculées sont donc des primes pures d'un même exercice.

| Paramètre | Valeur | Justification |

| Grille de marges | de 0 à 0,40, pas de 0,01 | Plage large qui contient l'optimum, au pas de 1 point de marge. |

Le prix proposé pour un profil dont la prime pure est $P$ est $\dfrac{P}{1 - \text{frais} - \text{marge}}$, de sorte qu'après paiement des sinistres et des frais il reste la marge cible. La probabilité de souscription décroît avec le rapport prix proposé sur prix de marché :

$$
p_{\text{sub}} = \text{plogis}\big(a - b\,(\text{ratio} - 1)\big)
$$

# Commentaire des graphiques et des résultats

## Graphique 1 : fréquence de sinistres par âge du conducteur

*Partie ciblée : analyse descriptive de la fréquence, avant le modèle de Poisson.*

Ce graphique montre la fréquence annuelle observée (sinistres par année d'exposition) pour chaque classe d'âge. Il traduit un fait central en assurance auto : le risque dépend fortement de l'âge. La fréquence des 17-20 ans dépasse 0,23 sinistre par an, celle des 21-25 ans est d'environ 0,13, puis elle tombe autour de 0,06 à 0,08 pour les classes plus âgées, avec une légère remontée entre 41 et 51 ans. Un jeune conducteur a donc en moyenne trois à quatre fois plus de sinistres qu'un conducteur d'âge moyen. Ce graphique justifie l'usage de l'âge comme variable de tarification et le découpage fin des classes jeunes. La classe `NA` est vide : elle correspond aux conducteurs de 100 ans exactement, exclus de la dernière classe `[71,100)` (voir remarques en fin de document).

## Tableau : fréquence par classe de bonus-malus

*Partie ciblée : analyse descriptive de la fréquence.*

La fréquence observée augmente de façon régulière avec le bonus-malus : environ 0,051 pour la classe (0,50], 0,075 pour (50,60], 0,112 pour (60,80], 0,143 pour (80,100] et 0,352 pour (100,150]. Le bonus-malus résume l'historique de sinistralité de l'assuré : un malus élevé signale un conducteur plus risqué. La classe (100,150] a une fréquence près de sept fois supérieure à la classe sans malus, ce qui en fait la variable la plus discriminante du portefeuille. Il est donc logique de l'inclure dans le modèle de fréquence.

## Modèle de fréquence (Poisson) et modèle de coût (Gamma)

*Partie ciblée : estimation des deux modèles sur le jeu d'apprentissage.*

Dans le modèle de Poisson, avec la classe 17-20 ans comme référence, les conducteurs de 26 à 30 ans ont une fréquence environ $e^{-0{,}56} \approx 0{,}57$ fois celle des plus jeunes. Un véhicule de plus de 15 ans a une fréquence environ $0{,}70$ fois celle d'un véhicule neuf, et une voiture à essence environ $0{,}84$ fois celle d'un diesel, à autres caractéristiques égales. Un véhicule de puissance 9 ou plus a une fréquence supérieure d'environ 30 % à la référence. Les effets de l'âge, du carburant et de la puissance sont très significatifs.

Dans le modèle Gamma, les variables explicatives agissent beaucoup moins sur le coût moyen : seuls certains effets de l'âge et de la zone sont significatifs. Cela traduit que **la tarification se joue surtout sur la fréquence** : le coût d'un sinistre dépend peu des caractéristiques de l'assuré, tandis que la probabilité d'en avoir un en dépend fortement.

## Graphique 2 : lift par décile de prime pure

*Partie ciblée : évaluation du modèle sur le jeu de test.*

Les contrats du jeu de test sont classés en 10 groupes de taille égale selon leur prime pure prédite (décile 1 = les moins risqués, décile 10 = les plus risqués). Pour chaque décile, la courbe « Prédit » donne la prime pure moyenne du modèle, et la courbe « Observé » la charge réelle par année d'exposition. Si le modèle est bon, les deux courbes sont proches et toutes deux croissantes.

C'est le cas ici. Les deux courbes montent ensemble, de 60 à 70 € par an dans le décile 1 jusqu'à environ 390 à 430 € dans le décile 10. Cela signifie que le modèle **classe correctement les risques** : les contrats jugés dangereux sont réellement ceux qui coûtent le plus cher. Quelques écarts apparaissent : le décile 8 est sous-estimé (l'observé dépasse le prédit) et le décile 10 est légèrement surestimé. Ces écarts sont modérés et normaux pour des déciles contenant peu de gros sinistres.

## Indicateurs de qualité : déviance, Gini, ratios par âge

*Partie ciblée : évaluation sur le jeu de test.*

- **Gini = 0,31.** Il mesure la capacité du modèle à séparer les bons et les mauvais risques : 0 correspond à un classement au hasard, 1 à un classement parfait. Un Gini d'environ 0,3 est correct pour un modèle de tarification auto sur ces données, où la sinistralité reste très aléatoire.
- **Déviance de Poisson.** Celle du modèle de référence constant (même fréquence pour tous) vaut environ 0,252. Le modèle complet doit faire mieux. Son calcul renvoie `NA` dans le notebook à cause des lignes à valeurs manquantes (voir remarques), et doit être refait après correction.
- **Ratios prédit / observé par classe d'âge.** Ils valent entre 0,97 et 1,03 pour presque toutes les classes, ce qui indique que le modèle est bien calibré. Le plus grand écart concerne les 71 ans et plus (1,09), classe de petite taille, donc plus bruitée.

## Graphique 3 : profit attendu selon la marge cible

*Partie ciblée : optimisation commerciale.*

Ce graphique donne le profit attendu total de l'assureur pour chaque marge cible comprise entre 0 et 40 %. Il traduit un arbitrage entre **marge par contrat** et **nombre de contrats vendus**. Quand la marge est faible, l'assureur vend beaucoup mais gagne très peu sur chaque contrat. Quand elle est forte, il gagne plus par contrat mais son prix dépasse celui du marché, donc les clients ne souscrivent plus. Le profit est donc nul pour une marge de 0, monte jusqu'à un maximum, puis redescend.

L'optimum se situe à une marge d'environ 13 % (0,13 par la grille, 0,131 par `optimize`), pour un profit attendu d'environ 1,49 million d'euros sur le jeu de test. Au-delà de 25 %, le profit chute fortement, car l'assureur perd presque tous ses clients. Ce résultat dépend entièrement des hypothèses fictives (prix de marché à 125 % de la prime pure, élasticité $a = 1$ et $b = 8$). Il illustre la méthode, il ne donne pas un prix réel.

# Interprétation et limites

**Ce que montrent les résultats.**

- Les variables de profil (âge, bonus-malus, ancienneté et puissance du véhicule) expliquent la fréquence de sinistres, ce qui justifie une tarification différenciée.
- Le modèle classe correctement les risques (Gini de 0,31, lift croissant) et reste bien calibré par classe d'âge.
- La marge optimale est un compromis entre rentabilité par contrat et volume de ventes.

**Limites.**

- Le plafonnement des gros sinistres à 181 252 € sous-estime les très grosses pertes, qu'il faudrait traiter à part.
- Le modèle de coût n'utilise pas toutes les variables du modèle de fréquence (ni bonus-malus, ni marque), et ses effets sont peu significatifs.
- Le modèle de fréquence ne tient pas compte d'une éventuelle surdispersion des données.
- L'hypothèse de non-corrélation entre fréquence et coût moyen n'est pas testée.
- Prix de marché et élasticité sont fictifs : le profit et la marge optimale ne sont pas utilisables tels quels.
- Pas de dimension temporelle (inflation, évolution du portefeuille).

# Remarques techniques sur le code

Ces points se corrigent dans le notebook avant de rendre les résultats définitifs.

1. **Valeurs manquantes sur l'âge.** `cut(DrivAge, breaks = c(17, 21, ..., 100), right = FALSE)` exclut la valeur 100 de la dernière classe, qui devient `NA`. C'est l'origine du `NA` dans le tableau par âge, des deux avertissements du graphique de lift, et de la déviance qui renvoie `NA`. Solution : utiliser `breaks = c(17, 21, 26, 31, 41, 51, 71, 101)`.
2. **Cellule 3.** Elle utilise `df` avant sa création, et transforme `DrivAgeClass` avant qu'elle existe. Elle doit être supprimée ou déplacée après la création des classes.
3. **Filtre des `NA`.** Il se trouve à la fin de la dernière cellule, après le calcul du profit. Il doit se faire avant, pour que les contrats à prime manquante n'influencent pas le résultat.
4. **Déviance de Poisson.** Après correction du point 1, elle se calcule normalement. Il faut aussi la calculer sur le même jeu pour le modèle complet et le modèle constant.
